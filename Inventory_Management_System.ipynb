```python
from abc import ABC, abstractmethod
import csv
import os

# 04. Abstraction: Abstract Base Class
class Item(ABC):
    @abstractmethod
    def display_info(self):
        pass

# 01. Classes & Objects & 03. Inheritance: Derived class inheriting from Item
class Product(Item):
    def __init__(self, product_id, name, price, quantity):
        # 02. Encapsulation: Private attributes
        self.__product_id = product_id
        self.__name = name
        self.__price = price
        self.__quantity = quantity

    # Getters and Setters for Encapsulation
    def get_product_id(self):
        return self.__product_id

    def set_product_id(self, product_id):
        self.__product_id = product_id

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price >= 0:
            self.__price = price
        else:
            raise ValueError("Price cannot be negative.")

    def get_quantity(self):
        return self.__quantity

    def set_quantity(self, quantity):
        if quantity >= 0:
            self.__quantity = quantity
        else:
            raise ValueError("Quantity cannot be negative.")

    def display_info(self):
        return f"ID: {self.__product_id} | Name: {self.__name} | Price: ${self.__price:.2f} | Qty: {self.__quantity}"

# 01. Classes & Objects: Manager class handling inventory operations
class InventoryManager:
    def __init__(self, filename="inventory.csv"):
        self.filename = filename
        self.products = []
        self.load_from_file() # 05. File Handling

    # 05. File Handling: Read data from CSV
    def load_from_file(self):
        try:
            if os.path.exists(self.filename):
                with open(self.filename, mode='r', newline='') as file:
                    reader = csv.reader(file)
                    for row in reader:
                        if row:
                            pid, name, price, qty = row[0], row[1], float(row[2]), int(row[3])
                            product = Product(pid, name, price, qty)
                            self.products.append(product)
        except Exception as e:
            print(f"Error reading file: {e}") # 06. Exception Handling

    # 05. File Handling: Write data to CSV
    def save_to_file(self):
        try:
            with open(self.filename, mode='w', newline='') as file:
                writer = csv.writer(file)
                for p in self.products:
                    writer.writerow([p.get_product_id(), p.get_name(), p.get_price(), p.get_quantity()])
        except Exception as e:
            print(f"Error saving to file: {e}") # 06. Exception Handling

    # Suggested Feature: Add products[cite: 1]
    def add_product(self, product_id, name, price, quantity):
        for p in self.products:
            if p.get_product_id() == product_id:
                print("Product ID already exists!")
                return
        try:
            new_product = Product(product_id, name, float(price), int(quantity))
            self.products.append(new_product)
            self.save_to_file()
            print("Product added successfully!")
        except ValueError as e:
            print(f"Invalid input: {e}") # 06. Exception Handling

    # Suggested Feature: Update products[cite: 1]
    def update_product(self, product_id, name=None, price=None, quantity=None):
        for p in self.products:
            if p.get_product_id() == product_id:
                try:
                    if name:
                        p.set_name(name)
                    if price is not None:
                        p.set_price(float(price))
                    if quantity is not None:
                        p.set_quantity(int(quantity))
                    self.save_to_file()
                    print("Product updated successfully!")
                    return
                except ValueError as e:
                    print(f"Error updating: {e}")
                    return
        print("Product not found.")

    # Suggested Feature: Delete products[cite: 1]
    def delete_product(self, product_id):
        for p in self.products:
            if p.get_product_id() == product_id:
                self.products.remove(p)
                self.save_to_file()
                print("Product deleted successfully!")
                return
        print("Product not found.")

    # Suggested Feature: Search products[cite: 1]
    def search_product(self, keyword):
        found = [p for p in self.products if keyword.lower() in p.get_name().lower() or keyword == p.get_product_id()]
        if found:
            print("\n--- Search Results ---")
            for p in found:
                print(p.display_info())
        else:
            print("No matching products found.")

    # Suggested Feature: Generate inventory reports[cite: 1]
    def generate_report(self):
        print("\n=== INVENTORY REPORT ===")
        if not self.products:
            print("Inventory is empty.")
            return
        total_value = 0
        for p in self.products:
            print(p.display_info())
            total_value += p.get_price() * p.get_quantity()
        print("-" * 30)
        print(f"Total Inventory Value: ${total_value:.2f}")

# Console Menu User Interface
def main():
    manager = InventoryManager()
    while True:
        print("\n--- INVENTORY MANAGEMENT SYSTEM ---")
        print("1. Add Product")
        print("2. Update Product")
        print("3. Delete Product")
        print("4. Search Product")
        print("5. Generate Report")
        print("6. Exit")

        choice = input("Enter choice (1-6): ")
        if choice == '1':
            pid = input("Enter Product ID: ")
            name = input("Enter Product Name: ")
            price = input("Enter Price: ")
            qty = input("Enter Quantity: ")
            manager.add_product(pid, name, price, qty)
        elif choice == '2':
            pid = input("Enter Product ID to update: ")
            name = input("Enter new Name (leave blank to skip): ")
            price = input("Enter new Price (leave blank to skip): ")
            qty = input("Enter new Quantity (leave blank to skip): ")
            manager.update_product(
                pid,
                name if name else None,
                price if price else None,
                qty if qty else None
            )
        elif choice == '3':
            pid = input("Enter Product ID to delete: ")
            manager.delete_product(pid)
        elif choice == '4':
            keyword = input("Enter Product ID or Name to search: ")
            manager.search_product(keyword)
        elif choice == '5':
            manager.generate_report()
        elif choice == '6':
            print("Exiting system. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 6.")

if __name__ == "__main__":
    main()
```
